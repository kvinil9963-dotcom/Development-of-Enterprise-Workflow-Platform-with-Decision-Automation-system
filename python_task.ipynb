 Qn1:Write a Python program to check whether a given number is prime.

 prompt:Write a Python program to check whether a given integer is prime. Explain the logic and show sample input and output.


```python

n = 29
is_prime = n > 1

for divisor in range(2, int(n ** 0.5) + 1):
    if n % divisor == 0:
        is_prime = False
        break

if is_prime:
    print(f"{n} is a prime number")
else:
    print(f"{n} is not a prime number")
```

Logic: A prime number is divisible only by 1 and itself.


Qn2:Write a Python program to reverse a string.

Prompt:Write a simple Python program to reverse a string.


```python

text = "Python"
reversed_text = text[::-1]

print("Original string:", text)
print("Reversed string:", reversed_text)
```

Logic: [::-1] slicing operation will reverse the string

```python

numbers = [12, 45, 7, 31, 26]
largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

Qn:4 Write a Python program to count the number of vowels in a string.

prompt:Write a simple Python program to count the vowels in a string, if user enters uppercase or lowercase letters ignore the differnces.


```python

text = "Hello World"
vowels = "aeiou"
count = 0

for character in text.lower():
    if character in vowels:
        count += 1

print("Number of vowels:", count)
```

Logic:checking each character in the vowels and incrementing the count

Qn5:Write a Python program to check whether a number is a palindrome.

prompt:Write a simple Python program to check whether a given number is a palindrome. provide sample input and output.

```python

n = 121
original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

if original == reverse:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```

Logic:Reverse the number and compare it with the original number

Qn6:Write a Python program to generate the Fibonacci series up to 10 terms.

prompt:Write a beginner-friendly Python program to generate the Fibonacci series for a given number of terms. show sample input and output.


```python

n = 10
a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

Logic:Each new term is the sum of the previous two terms


Qn7:Write a Python program to find the second-largest number in a list.

prompt:Write a beginner-friendly Python program to find the second-largest distinct number in a list without using the builtin function.

```python

numbers = [10, 20, 5, 40, 30]

largest = float("-inf")
second_largest = float("-inf")

for number in numbers:
    if number > largest:
        second_largest = largest
        largest = number
    elif number > second_largest and number != largest:
        second_largest = number

if second_largest == float("-inf"):
    print("No second-largest number")
else:
    print("Second-largest number:", second_largest)
```

Logic:Track the largest and second-largest distinct values while traversing the list


Qn8:Write a Python program to remove duplicate elements from a list.

prompt:Write a simple Python program to remove duplicate elements from a list while preserving the original order.

```python

numbers = [1, 2, 3, 2, 4, 1, 5]
unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

print("List without duplicates:", unique)
```

Logic:Add an element to the new list only if it is not already present


Qn9:Write a Python program to sort a list in ascending order without using the sort() function.

prompt:Write a simple Python program to sort a list in ascending order using the Bubble Sort algorithm without using sort() or sorted().

```python

numbers = [5, 2, 8, 1, 3]

n = len(numbers)

for i in range(n):
    for j in range(n - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Sorted list:", numbers)
```

Logic:Bubble Sort compares adjacent elements and swaps them if they are in the wrong order


Qn10:Write a Python program to find common elements in two lists.

prompt:Write a simple Python program to find the common elements between two lists without using set().

```python

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

common = []

for number in list1:
    if number in list2 and number not in common:
        common.append(number)

print("Common elements:", common)
```

Logic: Check whether each element in the first list also exists in the second list.
