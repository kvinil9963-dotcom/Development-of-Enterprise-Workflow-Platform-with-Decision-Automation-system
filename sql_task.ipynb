Qn1:Write an SQL query to display employees whose salary is greater than 50000.

prompt:Write a simple SQL query to select employees whose salary is greater than 50000. Use a sample employees table

```python

import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department TEXT,
    salary INTEGER
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?, ?)
""", [
    (101, "Asha", "IT", 60000),
    (102, "Kiran", "HR", 45000),
    (103, "Ravi", "IT", 75000),
    (104, "Meena", "Sales", 50000)
])

query = """
SELECT employee_id, name, salary
FROM employees
WHERE salary > 50000;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic: The WHERE clause filters employees whose salary is above 50000.


Qn2:Write an SQL query to find the average salary of employees.

prompt:Write an easy SQL query to calculate the average salary from an employees table.

```python

import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    salary INTEGER
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", 60000),
    (102, "Kiran", 45000),
    (103, "Ravi", 75000),
    (104, "Meena", 50000)
])

query = """
SELECT AVG(salary) AS average_salary
FROM employees;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic: AVG() calculates the average of all salary values.


Qn3:Write an SQL query to count the number of employees in each department.

prompt:Write a simple SQL query that counts employees in each department using GROUP BY. Explain the logic and show sample output.

```python

import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department TEXT
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", "IT"),
    (102, "Kiran", "HR"),
    (103, "Ravi", "IT"),
    (104, "Meena", "Sales")
])

query = """
SELECT department, COUNT(*) AS employee_count
FROM employees
GROUP BY department
ORDER BY department;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic: GROUP BY groups employees by department, and COUNT() counts employees in each group.

Qn5:Write an SQL query to display employee names along with their department names using INNER JOIN.

Prompt:Write a simple SQL INNER JOIN query to show employee names with their department names. Use sample employees and departments tables.

```python

import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department_id INTEGER
)
""")

connection.execute("""
CREATE TABLE departments (
    department_id INTEGER,
    department_name TEXT
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", 1),
    (102, "Kiran", 2),
    (103, "Ravi", 1)
])

connection.executemany("""
INSERT INTO departments VALUES (?, ?)
""", [
    (1, "Information Technology"),
    (2, "Human Resources")
])

query = """
SELECT employees.name, departments.department_name
FROM employees
INNER JOIN departments
ON employees.department_id = departments.department_id;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic: INNER JOIN combines rows from both tables where the department IDs match.


Qn6:Write an SQL query to retrieve the second highest distinct salary from the employees table.

prompt:Write a beginner-friendly SQL query to find the second highest distinct salary using a subquery and MAX() function.

```python
import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    salary INTEGER
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", 60000),
    (102, "Kiran", 45000),
    (103, "Anil", 75000),
    (104, "Meena", 50000),
    (105, "Rahul", 75000)
])

query = """
SELECT MAX(salary)
FROM employees
WHERE salary < (SELECT MAX(salary) FROM employees);
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic:The inner query SELECT MAX(salary) FROM employees finds the highest salary in the table.

The outer query uses WHERE salary < (SELECT MAX(salary) FROM employees) to exclude the highest salary.

The MAX(salary) function in the outer query finds the highest salary among the remaining salaries.

This gives the second highest distinct salary, even if multiple employees have the highest salary.

Qn8:Write an SQL query to retrieve all employees along with their department names, including employees who are not assigned to any department.

prompt:Write a beginner-friendly SQL query using LEFT JOIN to display all employees and their department names.

```python
import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE departments (
    department_id INTEGER,
    department_name TEXT
)
""")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    department_id INTEGER
)
""")

connection.executemany("""
INSERT INTO departments VALUES (?, ?)
""", [
    (1, "IT"),
    (2, "HR"),
    (3, "Finance")
])

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", 1),
    (102, "Kiran", 2),
    (103, "Anil", 1),
    (104, "Meena", None)
])

query = """
SELECT employees.name, departments.department_name
FROM employees
LEFT JOIN departments
ON employees.department_id = departments.department_id;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic:LEFT JOIN returns all rows from the employees table.

Matching department names are retrieved from the departments table.

If an employee has no matching department, the department name is displayed as NULL.


Qn9:Write an SQL query to retrieve employee names and their manager names using a self join.

prompt:Write a beginner-friendly SQL query using a self join to display employee names and their manager names.

```python
import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    manager_id INTEGER
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Rahul", None),
    (102, "Asha", 101),
    (103, "Kiran", 101),
    (104, "Anil", 102)
])

query = """
SELECT e.name AS employee_name,
       m.name AS manager_name
FROM employees e
LEFT JOIN employees m
ON e.manager_id = m.employee_id;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

logic:The employees table contains both employee and manager details.

The table is joined with itself using two aliases: e for employees and m for managers.

e.manager_id = m.employee_id matches each employee with their manager.

LEFT JOIN ensures employees without managers are also displayed.

Qn10:Write an SQL query to display each employee's name, salary, and salary status. If the salary is greater than or equal to 50000, display "High Salary"; otherwise, display "Low Salary".

prompt:Write a beginner-friendly SQL query to implement IF-ELSE logic using the CASE statement. Display employee name, salary, and salary status.

```python
import sqlite3

connection = sqlite3.connect(":memory:")

connection.execute("""
CREATE TABLE employees (
    employee_id INTEGER,
    name TEXT,
    salary INTEGER
)
""")

connection.executemany("""
INSERT INTO employees VALUES (?, ?, ?)
""", [
    (101, "Asha", 60000),
    (102, "Kiran", 35000),
    (103, "Anil", 75000),
    (104, "Meena", 45000)
])

query = """
SELECT name,
       salary,
       CASE
           WHEN salary >= 50000 THEN 'High Salary'
           ELSE 'Low Salary'
       END AS salary_status
FROM employees;
"""

result = connection.execute(query)

for row in result:
    print(row)

connection.close()
```

Logic:The CASE statement works like an IF-ELSE condition in SQL.

It checks whether the employee's salary is greater than or equal to 50000.

If the condition is true, it displays High Salary.

Otherwise, the ELSE clause displays Low Salary.

END finishes the CASE statement.
